```python
from time import time
from occsfrd import ansatz, wick, interface
```

```python
excitationLevels = [2]
BCH = ansatz.closedshellcc.genClosedShellCCAnsatz(excitationLevels)

equations = []

t0 = time()
equations.append(ansatz.closedshellcc.getEnergyEquation(BCH))
t1 = time()

print("Time to find energy equation:", t1 - t0)
print(equations[0])
print("number of terms:", len(equations[0].summandList))

equations.append(ansatz.closedshellcc.getAmplitudeEquation(BCH, excitationLevels[0]))

t2 = time()
print("Time to find doubles amplitude equation:", t2 - t1)
print(equations[1])
print("number of terms:", len(equations[1].summandList))
```

```python


interface.storeequations.save(equations, "CCDEquations")
interface.texify.texify(equations, "CCDEquations")
```
